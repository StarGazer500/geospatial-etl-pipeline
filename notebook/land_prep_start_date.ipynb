```python
import pandas as pd
import geopandas as gpd
```

```python
def read_data(path, sheet_name):
    data = pd.read_excel(path, sheet_name=sheet_name)
    
    data = data[data['QC Status'] == 'Pass']
    
    data = data[
        (data["Act.Code"].str.startswith("LP", na=False)) &
        (~data["Act.Name"].str.contains("Pitting for Blanking", case=False, na=False))
    ]
    
    data['WO_creation_date'] = pd.to_datetime(data['WO_creation_date'])
    data = data.sort_values(by="WO_creation_date", ascending=True)
    
    return data
```

```python
data = read_data(r"C:\Users\LENOVO\Downloads\Akwaaba work order tracker (1).xlsx",sheet_name="work_order_tracker")
```
